## Enunciado

En época invernal, es común que algunas ciudades experimenten lluvias intensas, lo que puede poner a prueba su infraestructura. Este año, ciudades como Valparaíso, Santiago y Concepción han registrado precipitaciones que requieren atención especial.


Durante un día de lluvia podemos representar la cantidad de agua acumulada, en milímetros (mm), en cada ciudad con respecto al tiempo transcurrido en horas (t) desde las 6 de la mañana, a través de las siguientes funciones cuadráticas:

**Valparaíso**: $$V(t)=0,7t^{2}+13,2t+39$$

**Santiago**: $$S(t) =1,1t^{2}+15,8t+32$$

**Concepción**: $$C(t) = 1,3t^{2}+18,9t+28$$

Se sabe que los sistemas de alcantarillado de las ciudades no pueden acumular más de 100 mm de agua. Si se supera este umbral, será necesario enviar equipos de emergencia para evitar inundaciones.

Con ayuda de Python responde las siguientes preguntas:

1. ¿Cuál es el valor de $t$ a las 07:30 de la mañana? ¿Cuál sería la cantidad de agua acumulada en Santiago a las 07:30 de la mañana?
1. En un mismo grafico, trace las tres funciones considerando que se pronostican 13 horas de lluvia.
1. ¿Cuál será la primera ciudad en requerir la ayuda de los equipos de emergencia?
1. Entre las 3 y 5 de la tarde, ¿en qué ciudad cayó más agua?

## Desarrollo

```python
#1
def V(t):
  imagen =0.7*t**2 +13.2*t +39
  return imagen

def S(t):
    imagen = 1.1*t**2 + 15.8*t + 32
    return imagen

def C(t):
  imagen = 1.3*t**2 + 18.9*t + 28
  return imagen

print("Para saber cuanto sera t debemos saber que el inicio es a las 6:00 y el que buscamos es a las 7:30 y la diferencia entre ambos es de 1:30 asi que t seria 1.5")
print("")
print(f"la cantidad que caera a las 7:30 sera de {S(1.5):.1f}mm")
```

```python
import matplotlib.pyplot as plt
import numpy as np

T = np.arange(0, 14, 1)

plt.xticks(np.arange(0, 14, 1))

val = V(T)
san = S(T)
con = C(T)

plt.plot(T, val, label = "Valparaíso", color = "red")
plt.plot(T, san, label = "Santiago", color = "blue")
plt.plot(T, con, label = "Concepción", color = "green")

plt.title("Lluvias torrenciales")
plt.xlabel("horas")
plt.ylabel("mm de agua")

plt.legend()
plt.grid(True)
plt.show()
```

```python
#3
from scipy.optimize import fsolve

#Val total
def total_1(t):
  return 0.7*t**2 +13.2*t - 70

#San total
def total_2(t):
    imagen = 1.1*t**2 + 15.8*t - 68
    return imagen

#Con total
def total_3(t):
  imagen = 1.3*t**2 + 18.9*t - 72
  return imagen


solucion_1 = fsolve(total_1, 3)
solucion_2 = fsolve(total_2, 3)
solucion_3 = fsolve(total_3, 3)

print(f"{solucion_1}")
print(f"{solucion_2}")
print(f"{solucion_3}")
print("")

print(f"La primera ciudad en requerir la ayuda de los equipos de emergencia es la ciudad de Concepción a las {solucion_1}")
print("")

total = f"""
    ¿Entre las 3 y 5 de la tarde, ¿en qué ciudad cayó más agua?

La ciudad que mas agua acumulo fue Concepción con {C(11) - C(9):.1f}MM
Santiago   = {S(11) - S(9):.1f}MM
Valparaíso = {V(11) - V(9):.1f}MM
"""
print(total)
```
