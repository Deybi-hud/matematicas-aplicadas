1. ¿Cuántas semanas deben transcurrir para que el ahorro alcance su valor máximo que pueden llegar a ahorrar?

```python
from scipy.optimize import fsolve


def a(n):
    return 50*n+50-18300

solucion = fsolve(a,10)[0]
print(f"Para llegar al maximo 18.300 deben pasar semanas {solucion:.0f}")
```

1. Los isleños han descubierto otro método de ahorro llamado "Ahorro Progresivo". Durante las primeras $10$ semanas, el monto ahorrado en la $n$-ésima semana se duplica con respecto a la semana anterior. Si en la primera semana ahorran $30, encuentra la expresión algebraica $P_{n}$ que represente el monto ahorrado en la $n$-ésima semana, e imprime la cantidad ahorrada durante las primeras $10$ semanas.

```python
def p(n):
    return 30*2**(n*1)

x = 0
for i in range(10):
    print(f"Semanas {i + 1} --> {p(i)}")
    x = p(i) + x

print(f"Total --> {x}")

```
