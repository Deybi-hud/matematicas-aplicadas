La empresa de Ciberseguridad Seguri-net está analizando dos métodos de encriptación para archivos desde $10$ MB hasta $80$ MB. En el primer método, por cada incremento de $2$ MB en el tamaño del archivo, el tiempo de encriptación se incrementa de manera constante en $0,8$ segundos, tal como muestra la tabla (Método A). Mientras que el segundo método (Método B) tiene un tiempo de encriptación, también en segundos, que varía de acuerdo a un modelo cuadrático.

La función que nos permite calcular el timepo de encriptación es una función afín ya que el método A dice que el tiempo de encriptación se **incrementa de manera constante**
 
$$m = 0.8 / 2 = 0.4$$

$$n = 7.4 - (0.4 * 10) = 3.4$$

$$f(x) = 0.4x + 3.4$$

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve


def f(x):
    imagen = 0.4*x + 3.4
    return imagen

def g(x):
    imagen = 0.01*x**2 - 0.25*x + 10
    return imagen

resultado_1 = f(27)
resultado_2 = g(27)

print(f"El resultado cuando se calcula con 27MB en cada método es de {resultado_1:.2f}s (metodo A) y {resultado_2}s (metodo B)") 
print(f"El método que empleó menos tiempo fue el método B con {resultado_2:.2f}s")   
```

```python

def interseccion(x):
    return 0.01*x**2 - 0.25*x + 10 - 0.4*x - 3.4

x = np.linspace(10, 80, 2)
solucion = fsolve(interseccion, x)
unica = np.unique(solucion)
print(f" {unica}")
print("")
resultado_3 = f(12.59397077)
resultado_4 = g(52.40602924)
#
#¿Para qué tamaño(s) de archivo el tiempo de encriptación es el mismo en ambos métodos? ¿Cuál es el tiempo de encriptación?

print(f"solo para 2 tamaños 12,59MB y 52,40MB en cual el tiempo de cada uno es {resultado_3:.2f}s y para el 2do {resultado_4:.2f}s")
```

```python
X = np.arange(10, 80, 10)
plt.xticks(np.arange(10, 80, 10))

Y = f(X)

G = g(X)

plt.plot(X,Y, label = 'f(x)', color= 'r')
plt.plot(X,G, label = 'G(x)', color= 'b')
plt.title('')
plt.xlabel('megabyte') 
plt.ylabel('tiempo')
plt.legend() 
plt.grid(True)
plt.show()
```
