Pregunta 1

```python
import matplotlib.pyplot as ptl
import numpy as np
from scipy.optimize import fsolve

def s(x):
    return -1.6*x**2+1.3*x+141.3

def r(x):
    return 1.4*x**2-5.6*x+12.6

def interseccion(x):
    return s(x)-r(x)

x = np.linspace(0,100,5)
solucion = np.around(fsolve(interseccion, x),1)
unique = np.unique(solucion)
print(f"{unique}")
```
