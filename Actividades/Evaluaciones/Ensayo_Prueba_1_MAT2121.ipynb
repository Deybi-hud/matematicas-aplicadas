$5) \space $ Imagina que trabajas en el área de operaciones de una empresa de energía renovable y estás analizando la producción acumulada de dos plantas solares que abastecen a una red eléctrica.La Planta Solar A genera energía acumulada de acuerdo con la función $$P(t) = 0,15t^2$$, donde $P(t)$ representa la cantidad de megavatios-hora (MWh) generados después de $t \color {red} $ meses.
La Planta Solar B, por su parte, genera energía según la función $$Q(t) = 0,10t^2 + 0,2t + 1,6$$, donde $Q(t)$ representa la cantidad de megavatios-hora (MWh) generados después de $t$ meses.<br><br>Encuentra el momento en que ambas plantas habrán generado la misma cantidad de energía.


```python
def p(t):
    return 0.15*t**2

def q(t):
    return 0.10*t**2+0.2*t+1.6

def inter(x):
    return p(x)-q(x)

solucion = fsolve(inter, 3)[0]
print(f"{round(solucion)}")

print(f"{p(8)}")
print(f"{q(8)}")
```

$6) \space $ Una empresa que fabrica y vende bicicletas eléctricas obtiene ingresos según la función:$$I(x) = 100x + 2x^2$$y sus costos operativos se modelan con la función:$$C(x) = 50x + 3500$$donde $x$ representa la cantidad de bicicletas fabricadas y vendidas, y tanto el ingreso como el costo están expresados en dólares.

$10) \space $ Una empresa requiere contratar licencias de un software de diseño para sus empleados. El Proveedor 1, donde no cobran costos fijos de implementación ni soporte, ofrecen cada licencia en 120 dólares, mientras que el Proveedor 2 ofrece la misma licencia en 70 dólares, considerando en este caso un costo fijo único de configuración e instalación del sistema de 300 dólares. Las funciones que permiten calcular el costo total (en dólares) con cada proveedor respectivamente, en función de la cantidad $x$ de licencias contratadas, son: $$f(x) = 120x$$ $$g(x) = 70x + 300$$El máximo de licencias analizadas en la evaluación corporativa es de 12 unidades.

a) Presente en un mismo gráfico las funciones que relacionan el costo con cada proveedor y la cantidad de licencias médicas.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 12, 12)

f_x = 120 * x           
g_x = 70 * x + 300      

plt.plot(x, f_x, label='Proveedor 1: f(x) = 120x', color='blue', marker='o')
plt.plot(x, g_x, label='Proveedor 2: g(x) = 70x + 300', color='orange', marker='s')

plt.title('Comparación de Costos de Licencias')
plt.xlabel('Cantidad de licencias (x)')
plt.ylabel('Costo total en dólares')
plt.axvline(x=6, color='red', linestyle='--', alpha=0.5, label='Punto de equilibrio (x=6)')
plt.legend()
plt.grid(True)
plt.show()
```
