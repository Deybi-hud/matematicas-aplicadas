# Ejercicio de Práctica 1: Consumo Eléctrico durante la Ola de Calor

Durante los meses de verano, las olas de calor provocan un aumento considerable en el uso de sistemas de aire acondicionado y refrigeración, poniendo a prueba la red eléctrica municipal. Este año, los distritos Norte, Centro y Sur han registrado una demanda energética que requiere supervisión continua.  Durante un día crítico de verano, el consumo acumulado de energía, en megavatios-hora ($MWh$), en cada distrito con respecto al tiempo transcurrido en horas ($t$) desde las 8:00 de la mañana, se representa mediante las siguientes funciones cuadráticas:

 Distrito Norte:$$N(t) = 1.5t^2 + 8.5t + 40$$Distrito Centro:$$C(t) = 2.0t^2 + 6.0t + 50$$Distrito Sur:$$S(t) = 1.2t^2 + 11.0t + 35$$


Se sabe que la infraestructura eléctrica de cada distrito tiene un límite máximo de capacidad acumulada de 150 MWh. Si el consumo supera este umbral, se activa automáticamente una alerta de sobrecarga que requiere la intervención de cuadrillas de soporte técnico.  Con ayuda de Python, responde las siguientes preguntas:  


1. ¿Cuál es el valor de $t$ a las 10:30 de la mañana?
2. ¿Cuál sería la cantidad de energía acumulada en el Distrito Sur a las 10:30 de la mañana?  En un mismo gráfico, trace las tres funciones considerando que el monitoreo dura un periodo total de 10 horas de calor intenso.  
3. ¿Cuál será el primer distrito en activar la alerta de sobrecarga y requerir la ayuda de las cuadrillas de soporte técnico?  Entre las 2 y las 4 de la tarde (14:00 y 16:00 h),
4. ¿qué distrito tuvo el mayor incremento/consumo de energía?
