# Ejercicio de Práctica 1: Consumo Eléctrico durante la Ola de Calor

Durante los meses de verano, las olas de calor provocan un aumento considerable en el uso de sistemas de aire acondicionado y refrigeración, poniendo a prueba la red eléctrica municipal. Este año, los distritos Norte, Centro y Sur han registrado una demanda energética que requiere supervisión continua.  Durante un día crítico de verano, el consumo acumulado de energía, en megavatios-hora ($MWh$), en cada distrito con respecto al tiempo transcurrido en horas ($t$) desde las 8:00 de la mañana, se representa mediante las siguientes funciones cuadráticas:

 Distrito Norte:$$N(t) = 1.5t^2 + 8.5t + 40$$Distrito Centro:$$C(t) = 2.0t^2 + 6.0t + 50$$Distrito Sur:$$S(t) = 1.2t^2 + 11.0t + 35$$


Se sabe que la infraestructura eléctrica de cada distrito tiene un límite máximo de capacidad acumulada de 150 MWh. Si el consumo supera este umbral, se activa automáticamente una alerta de sobrecarga que requiere la intervención de cuadrillas de soporte técnico.  Con ayuda de Python, responde las siguientes preguntas:  


1. ¿Cuál es el valor de $t$ a las 10:30 de la mañana?
2. ¿Cuál sería la cantidad de energía acumulada en el Distrito Sur a las 10:30 de la mañana?  En un mismo gráfico, trace las tres funciones considerando que el monitoreo dura un periodo total de 10 horas de calor intenso.  
3. ¿Cuál será el primer distrito en activar la alerta de sobrecarga y requerir la ayuda de las cuadrillas de soporte técnico?  Entre las 2 y las 4 de la tarde (14:00 y 16:00 h),
4. ¿qué distrito tuvo el mayor incremento/consumo de energía?

```python
print("El valor de t a las 10:30 es de 2.5")

print("") 


def calculo_1(t):
    imagen = 1.5*t**2 + 8.5*t + 40 
    return imagen

def calculo_2(t):
    imagen = 2.0*t**2 + 6.0*t + 50
    return imagen

def calculo_3(t):
    imagen = 1.2*t**2 + 11.0*t + 35
    return imagen

resultado_3 = calculo_3(2.5)

total = f"""El resultado de la cantidad de energía generada a las 10:30 es de {resultado_3:.0f}MWh"""

print(total)
```

```python
import matplotlib.pyplot as plt
import numpy as np

T = np.arange(0, 11, 1)

plt.xticks(np.arange(0, 11, 1))

norte = calculo_1(T)
centro = calculo_2(T)
sur = calculo_3(T)

plt.plot(T, sur, label="Distrito Sur", color = "blue")
plt.plot(T, norte, label="Distrito Norte", color = "red")
plt.plot(T, centro, label="Distrito Centro", color = "green")

plt.title("Generación de Energía por Distrito")

plt.xlabel("Tiempo (horas)")
plt.ylabel("Cantidad de energía generada (MWh)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
norte_resultado = calculo_1(6)
centro_resultado = calculo_2(6)
sur_resultado = calculo_3(6)

resultado_total = f"""
El resultado de la cantidad de energía generada del Distrito norte a las 14:00 es de {norte_resultado}MWh
El resultado de la cantidad de energía generada del Distrito centro a las 14:00 es de {centro_resultado}MWh
El resultado de la cantidad de energía generada del Distrito sur a las 14:00 es de {sur_resultado}MWh
"""

print(resultado_total)

print(f"El primer Distrito en activar la alerta de sobre carga es Distrito Centro dando un resutado a las 14:00 de {centro_resultado} ")
print(f"El que tuvo mas consumo fue el Distrito Centro")
```

# Ejercicio de Práctica 2: Gestión de Residuos Urbanos en Temporada Festiva

Durante las festividades de fin de año, el aumento del comercio y las celebraciones genera un incremento masivo en la recolección de residuos sólidos en los sectores Oriente, Poniente y Sur de la ciudad.El volumen acumulado de residuos recolectados en toneladas ($t$), en función del tiempo transcurrido en horas ($h$) desde las 7:00 de la mañana, se modela mediante las siguientes funciones lineales y cuadráticas:

Sector Oriente (Modelo cuadrático):$$O(h) = 0,6h^2 + 7,5h + 15$$Sector Poniente (Modelo lineal):$$P(h) = 14,2h + 10$$Sector Sur (Modelo cuadrático):$$S(h) = 0,9h^2 + 5,0h + 20$$Los centros de acopio temporal de cada sector tienen una capacidad máxima de 110 toneladas acumuladas. Si un sector supera este umbral, el centro de acopio entra en saturación y requiere el despacho de camiones de carga pesada de refuerzo. Con ayuda de Python, responde las siguientes preguntas:


1. ¿Cuál es el valor de $h$ a las 11:30 de la mañana?
2. ¿Cuál sería la cantidad de residuos acumulados en el Sector Poniente a las 11:30 de la mañana?En un mismo gráfico, trace las tres funciones considerando un turno continuo de 9 horas de recolección.
3. ¿Cuál será el primer sector en entrar en saturación y requerir camiones de refuerzo?Entre las 1:00 p.m. y las 3:00 p.m. (13:00 y 15:00 h),
4. ¿qué sector acumuló la mayor cantidad de residuos?

```python
print("El valor de h a las 11:30 es de 4.5")

print("")

def calculo_4(h):
    imagen = 0.6*h**2 + 7.5*h + 15
    return imagen

def calculo_5(h):
    imagen = 14.2*h + 10
    return imagen

def calculo_6(h):
    imagen = 0.9*h**2 + 5.0*h + 20
    return imagen


resultado_4 = calculo_5(4.5)

print(f"La cantidad de residuos acumulados en el Sector Ponientes a las 11:30 es de {resultado_4} toneladas")
```

```python
import matplotlib.pyplot as plt
import numpy as np


X = np.linspace(0, 9, 100)

plt.xticks(np.arange(0, 10, 1))

Oriente = calculo_4(X)

Ponientes = calculo_5(X)

Sur = calculo_6(X)

plt.plot(X, Oriente, label="Sector Oriente", color = "red")
plt.plot(X, Ponientes, label="Sector Ponientes", color = "blue")
plt.plot(X, Sur, label="Sector Sur", color = "orange")

plt.title("Acumulación de Residuos en el los 3 Sectores")
plt.xlabel("Tiempo (horas)")
plt.ylabel("Cantidad de residuos acumulados (toneladas)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
Oriente = calculo_4(8)
Ponientes = calculo_5(8)
Sur = calculo_6(8)

total_residuos = f"""
El resultado {Oriente}
El resultado {Ponientes}
El resultado {Sur}
"""

print(total_residuos)

print(f"El primer Sector en entrar saturación es el Sector Poniente dando un resultado de {Ponientes}")
print("")
print("El Sector que acumulo mas residuos fue el sector Poniente")
```
